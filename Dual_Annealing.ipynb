```python
import numpy as np
from scipy.spatial.distance import pdist
from scipy.optimize import dual_annealing
import plotly.graph_objects as go

# Lennard-Jones potential function
def lj_potential(r, epsilon=1.0, sigma=1.0):
    term1 = (sigma / r) ** 12
    term2 = (sigma / r) ** 6
    return 4 * epsilon * (term1 - term2)

# Objective function: Total Lennard-Jones potential for N particles in 3D
def total_lj_energy(coords, N, box_size):
    coords = coords.reshape((N, 3))  # Reshape flat array into (N, 3)

    # Compute pairwise distances in 3D
    distances = pdist(coords)  # Condensed pairwise distance matrix

    # Sum Lennard-Jones potential for each pair
    energy = np.sum(lj_potential(distances))

    return energy

# Number of particles
N = 7  # Number of particles
box_size = 5.0  # Box dimensions
num_restarts = 5  # Number of restarts

# Store the best result
best_result = None

# Run multiple restarts
for _ in range(num_restarts):
    # Initial positions (randomly placed in a cubic box)
    x0 = np.random.uniform(0, box_size, size=(N, 3)).flatten()

    # Bounds: Keep particles within the box
    bounds = [(0, box_size) for _ in range(3 * N)]

    # Run Simulated Annealing (Dual Annealing)
    result = dual_annealing(total_lj_energy, bounds=bounds, args=(N, box_size))

    # Store the best result
    if best_result is None or result.fun < best_result.fun:
        best_result = result

# Extract optimized positions
optimized_coords = best_result.x.reshape((N, 3))

# Print results
print("Optimized Positions:\n", optimized_coords)
print("Minimum Energy:", best_result.fun)

# Create interactive 3D scatter plot using Plotly
fig = go.Figure()

fig.add_trace(go.Scatter3d(
    x=optimized_coords[:, 0],
    y=optimized_coords[:, 1],
    z=optimized_coords[:, 2],
    mode='markers',
    marker=dict(size=6, color='red', opacity=0.8),
    name="Optimized Particles"
))

# Set axis labels
fig.update_layout(
    title="Optimized Particle Positions (Lennard-Jones Potential)",
    scene=dict(
        xaxis_title="X",
        yaxis_title="Y",
        zaxis_title="Z",
    ),
    margin=dict(l=0, r=0, b=0, t=40)
)

# Show interactive plot
fig.show()
```
